# 📊 DeepGuardDB Training Dashboard
Run this notebook to see the current progress of your AI detection models.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

def plot_metrics(csv_path, title):
    if not Path(csv_path).exists():
        print(f"Waiting for {csv_path} to be created...")
        return
    
    df = pd.read_csv(csv_path)
    if df.empty:
        print("No data yet.")
        return

    plt.figure(figsize=(12, 4))
    
    # Subplot 1: Loss
    plt.subplot(1, 2, 1)
    plt.plot(df['epoch'], df['train_loss'], label='Train Loss', marker='o')
    plt.plot(df['epoch'], df['val_loss'], label='Val Loss', marker='x')
    plt.title(f'{title} - Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True, alpha=0.3)

    # Subplot 2: Accuracy
    plt.subplot(1, 2, 2)
    plt.plot(df['epoch'], df['train_acc'], label='Train Acc', marker='o', color='green')
    plt.plot(df['epoch'], df['val_acc'], label='Val Acc', marker='x', color='orange')
    plt.title(f'{title} - Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()
    plt.grid(True, alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"Latest Val Acc: {df['val_acc'].iloc[-1]:.2%}")
```

## 1. Model A: Spatial CNN
This was the first model we trained.

```python
plot_metrics('logs/cnn_spatial_metrics.csv', 'Spatial CNN')
```

## 2. Model B: FFT CNN
This is the model currently being refined.

```python
plot_metrics('logs/cnn_fft_metrics.csv', 'FFT CNN')
```
